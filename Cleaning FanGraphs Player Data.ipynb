```python
import pandas as pd

file_path = '/content/Fangraphs Data for DATA 606 - Batting 2016-2025.csv'
df = pd.read_csv(file_path)

# Display the first 5 rows and the column names to verify
print(df.head())
print(df.columns)
```

```python
df.info()
```

Let's check the null values in "Fld" first.

```python
df[df['Fld'].isnull()]
```

These are all players with limited time in the field, due to being Designated Hitters. As such, we should impute a 0 for the nulls.

```python
df['Fld'] = df['Fld'].fillna(0)

# Display the first few rows to verify the imputation
print(df[df['Fld'] == 0].head())
```

The only other columns with null values are "UBR" and "wGDP". These are both related to a player's baserunning ability. Since "XBR" is another metric for baserunning ability, and has no null values, we can safely drop these two columns.

```python
df = df.drop(columns=['UBR', 'wGDP'])
df.info()
```

We have removed all the null values from the dataset. We can now sort it. Let's sort by player and then year.

```python
df_sorted = df.sort_values(by=['PlayerId', 'Season'])
print(df_sorted.head())
```
